```python
# Block 1: Define Variables & Constants
HOSPITAL_NAME = "Z Medical Center"

# Defining variables
patient_name = ""  # String
patient_age = 0  # Int
patient_weight = 0.0  # Float
patient_height = 0.0  # Float

# Block 2: Collect Keyboard Input
patient_name = input("Enter patient name: ")
patient_age = int(input("Enter patient age: "))
patient_weight = float(input("Enter patient weight (kg): "))
patient_height = float(input("Enter patient height (m): "))

# Block 3: Print Formatted Patient Summary
print("\n📋 Patient Summary")
print(f"👤 Name: {patient_name}")
print(f"🎂 Age: {patient_age}")
print(f"⚖️ Weight: {patient_weight:.2f} kg")
print(f"📏 Height: {patient_height:.2f} m")
print(f"🏥 Hospital: {HOSPITAL_NAME}")

# Block 4: Decision Structure - if Statement
if patient_age >= 65:
    print("🏅 This patient qualifies as a senior citizen.")

# Block 5: Decision Structure - if-else Statement
if patient_age >= 65:
    print("✅ Eligible for senior healthcare benefits.")
else:
    print("❌ Not eligible for senior healthcare benefits.")

# Block 6: Decision Structure - if-elif-else Statement
if patient_age < 18:
    print("🧒 This patient is a minor.")
elif 18 <= patient_age <= 64:
    print("🧑 This patient is an adult.")
else:
    print("👴 This patient is a senior citizen.")

# Block 7: Boolean Variables & Logical Operators
has_insurance = input("Does the patient have insurance? (yes/no): ").strip().lower() == "yes"

if patient_age >= 65 and has_insurance:
    print("🏥 Senior citizen with insurance.")
elif patient_age >= 65 and not has_insurance:
    print("⚠️ Senior citizen without insurance. Need additional support.")
else:
    print("🔄 General patient processing.")
```
